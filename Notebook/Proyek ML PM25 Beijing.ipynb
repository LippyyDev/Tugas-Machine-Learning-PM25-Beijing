## 02. Muat data dan EDA

```python
from pathlib import Path

# Notebook dijalankan dari folder Notebook/
DATA_DIR = Path('../Data/Data Mentah')

berkas = sorted(DATA_DIR.glob('*.csv'))
print(len(berkas), 'berkas stasiun')

df = pd.concat([pd.read_csv(f) for f in berkas], ignore_index=True)
df['datetime'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])
df = df.sort_values(['station', 'datetime']).reset_index(drop=True)
print(df.shape)
df.head()
```

```python
# Pemeriksaan struktur dasar
print(df.groupby('station').size())
print('Rentang waktu:', df['datetime'].min(), 'sampai', df['datetime'].max())
print('Duplikat (stasiun, waktu):', df.duplicated(['station', 'datetime']).sum())
(df.isna().mean() * 100).round(2)
```
